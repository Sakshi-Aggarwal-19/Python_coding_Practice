1.Reverse a string without using slicing.


```python
name = "Sakshi"

rev = " "

for i in name:
  rev = i + rev

print(rev)


```

2. Check if a string is a palindrome.

```python
string = "MadaM"

rev_str = ""

for i in string:
  rev_str = i + rev_str


if string == rev_str:
  print("Palindrome")
else:
  print("Not palindrome")
```

3.Count vowels and consonants in a string.

```python
check_num = "jfhsdulaidkajdw"

vowel_count = 0
consonent_count = 0

Vowel = 'aeiouAEIOU'

for i in check_num:
  if i in Vowel:
    vowel_count +=1
  else:
    consonent_count +=1


print(vowel_count)
print(consonent_count)
```

4.Find duplicate characters in a string.

```python
str2 = "nhsjfhwkduksjasnx"


for i in str2:
  count = 0

  for j in str2:
    if i == j:
      count +=1

  if count>1:
      print(i)

```

5.Find the first non-repeating character.

```python
name = "fsfhskfksdladud"

# first non-repeating character
for i in name:
  count = 0
  for j in name:
    if i ==j:
      count +=1

  if count ==1:
    print(i)
    break
```

6.Find the first repeating character.

```python
name = "zsfhskfksdladud"

# first repeating character
for i in name:
  count = 0
  for j in name:
    if i ==j:
      count +=1

  if count > 1:
    print(i)
    break
```

8.Count frequency of each character.

```python
name2 = "mbjhdfsydisudsnd"

freq={}

for ch in name2:
  if ch in freq:
    freq[ch] = freq[ch] + 1
  else:
    freq[ch] = 1

print(freq)
```

9.Check if two strings are anagrams.

```python
str1 = "tab"
str2 = "tab"

if sorted(str1) == sorted(str2):
  print("anagrams")
else:
  print("not anagrams")
```

10.Reverse the order of words in a sentence.

```python
sentence = "My name is Sakshi Aggarwal"

sentence = sentence.split(' ')
print(sentence)

rev_sen = " "

for i in sentence:
  rev_sen = i+ " " + rev_sen

print(rev_sen)
```
